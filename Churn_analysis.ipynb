```python
import pandas as pd

df_tickets = pd.read_csv('/Users/prachisharma/Desktop/Churn Analysis Project/ravenstack_support_tickets.csv')

df_tickets.head(10)
```

```python
df_tickets.info()
```

```python
df_tickets['submitted_at'] = pd.to_datetime(df_tickets['submitted_at'])
df_tickets['closed_at'] = pd.to_datetime(df_tickets['closed_at'])
```

```python
df_tickets['priority'] = df_tickets['priority'].str.strip().str.lower()
```

```python
df_tickets['escalation_flag'] = df_tickets['escalation_flag'].astype(int)
```

```python
df_tickets.info()
```

```python
df_tickets[df_tickets['satisfaction_score'].isnull()]
```

```python
# fill median score to the missing values

median_score = df_tickets['satisfaction_score'].median()
df_tickets['satisfaction_score'] = df_tickets['satisfaction_score'].fillna(median_score)
```

```python
df_tickets['response_category'] = pd.cut(
    df_tickets['first_response_time_minutes'],
    bins=[0, 30, 60, 120, 180],
    labels=['fast', 'moderate', 'slow', 'very slow']
)

df_tickets.to_csv('ravenstack_support_tickets_cleaned.csv', index=False)
```
